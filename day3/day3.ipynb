```python
class daughter:
    def __init__(self, name, id):
        self.name = name
        self.id = id

    def showDetails(self):
        print(f"The name of daughter is {self.name}, id is {self.id}")


class father(daughter):
    def showLanguage(self):
        print("Father is speaking Bengali language")


class grandchild(daughter):
    def showGeneration(self):
        print("I am the grandchild")


d1 = daughter("Riya", 1)
d1.showDetails()

d2 = father("Madhu", 10)
d2.showDetails()
d2.showLanguage()

d3 = grandchild("Punam", 20)
d3.showDetails()
d3.showGeneration()
```

```python
class student:
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks
    def display(self):
        print(f"the name of student is {self.name} and marks is {self.marks}")
class CollegeStudent(student):
     def __init__(self,name,marks,course):
        super().__init__(name,marks)
        self.course=course 
     def display(self):
        print(f"the name of student is {self.name} and marks is {self.marks} and course is {self.course}")

a=student("madhu",90)
a.display()
b=CollegeStudent("punam",80,"ML")
b.display()

    
    
```
