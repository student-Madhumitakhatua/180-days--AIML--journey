🟢Question 1

 
Create a CSV file named students.csv with columns Name, Age, and Marks, then read it using Pandas.

```python
import pandas as pd
MK=pd.read_csv("students.csv")
MK.head()
```

🟢Question 2
 
Display only the Marks column from the DataFrame.

```python
MK["Marks"]
```

🟢Question 3

 
Display only the students whose marks are greater than 80.

```python
MK[MK["Marks"]>80]
```

🟢Question 4

 
Display students whose marks are greater than or equal to 85.

```python
MK[MK["Marks"]>=85]
```

🟢 Question 5


Display students whose Age is greater than 21.

```python
MK[MK["Age"]>21]
```

Question 6

 
Display students whose Marks are greater than 80 AND Age is greater than 21.

```python
MK[(MK["Marks"] > 80) & (MK["Age"] > 21)]
```

 🟢 Question 7

 
Display students whose Marks are less than 80 OR Age is less than 21.

```python
MK[(MK["Marks"]<80) | ( MK ["Age"] < 21)]
```

Question 8

 
Display only the Name and Marks columns for students whose marks are greater than 80.

```python
MK[MK["Marks"] > 80] 
MK[["Name", "Marks"]]
```

Question 9
 
Sort the DataFrame by Marks from highest to lowest.

```python
MK.sort_values("Marks", ascending=True)
```

```python
MK.sort_values("Marks", ascending=False)
```

Question 10 — Final Practice

 
Find the student with the highest marks and display their complete row.

```python
MK.sort_values("Marks",ascending =False).iloc[0]
```
