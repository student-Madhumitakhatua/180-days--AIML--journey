#MAP

```python
l=[1,2,4,5,6,8,9]
newl=list(map(lambda x: x*x*x,l))
print(newl)
```

```python
numbers = [2, 4, 6, 8, 10]
new_numbers=list(map(lambda x:x*2,numbers))
print(new_numbers)
```

```python
numbers = [3, 5, 7, 10]
squarelist=[]
for x in numbers:
    squarelist.append(pow(x,2))
print(squarelist)
```

```python
numbers = [3, 5, 7, 10]
def square_me(x):
    return x*x
squareList=map(square_me,numbers)
print(list(squareList))
```

```python
numbers = [3, 5, 7, 10]
squarelist=map(lambda x: x*x,numbers)
print(list(squarelist))
```

#FILTER

```python
numbers = [1, 2, 3, 4, 5, 6]
even=list(filter(lambda x:x%2==0,numbers))
print(even)
```

```python
numbers = [5, 12, 8, 20, 3, 15, 7]
def filter_function(a):
    return a>10
filter_numbers=list(filter(filter_function,numbers))
print(filter_numbers)
```

```python
numbers = [5, 12, 8, 20, 3, 15, 7]
 
filter_numbers=list(filter(lambda x: x>10, numbers))
print(filter_numbers)
```

```python
numbers = [4, 12, 7, 18, 9, 22, 15, 30]
filter_numbers=list(filter(lambda x:x>10 and x%2==0,numbers))
print(filter_numbers)
```

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
filter_numbers=list(map(lambda x:x*x,filter(lambda x:x%2==0,numbers)))
print(filter_numbers)
```

Problem 2 — Square Numbers Greater Than 10

```python
numbers = [4, 8, 12, 15, 3, 20]
square=list(map(lambda x: x*x,filter(lambda x: x>10,numbers)))
print(square)
```

Problem 3 — Odd Numbers Cube

```python
numbers = [2, 5, 8, 11, 14, 17, 20, 23]
cube=list(map(lambda x:x*x*x,filter(lambda x: x%2!=0,numbers)))
print(cube)
```

Mini challenge

```python
students = [
    {"name": "A", "score": 75},
    {"name": "B", "score": 85},
    {"name": "C", "score": 90},
    {"name": "D", "score": 68},
    {"name": "E", "score": 82}
]
high_score=list(map(lambda student: student["score"] + 5, filter(lambda student: student["score"] >= 80, students)))
print(high_score)
```
