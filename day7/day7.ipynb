Question 2 — Python + Dictionary

Find and print the student with the highest marks

```python
students = {
    "Madhu": 85,
    "Punam": 72,
    "Rahul": 90,
    "Amit": 65
}

print(max(students.values()))

for name,marks in students.items():
    if marks == max(students.values()):
        print(name,marks)
```

Question 3 — Functions

Write a function called calculate_average() that takes a list of numbers and returns the average.

```python
 
def calculate_average(numbers):
    return  sum(numbers)/len(numbers)
numbers = [10, 20, 30, 40, 50]
 
print(calculate_average(numbers))
```

Question 4 — OOP


```python
class Student:
    def __init__(self, name, marks):
        self.name=name
        self.marks=marks
    def display(self):
        print(f" name: {self.name}  marks: {self.marks}")

Student1=Student("Madhu", 90)
Student1.display()
```
