```python
import sqlite3

conn = sqlite3.connect("day19.db")
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE students (
    Name TEXT,
    Age INTEGER,
    Marks INTEGER,
    Course TEXT
)
""")

students_data = [
    ("Madhu", 25, 90, "AI"),
    ("Punam", 20, 80, "Web"),
    ("Rahul", 22, 85, "ML"),
    ("Amit", 21, 70, "AI"),
    ("Priya", 23, 95, "ML")
]

cursor.executemany("""
INSERT INTO students (Name, Age, Marks, Course)
VALUES (?, ?, ?, ?)
""", students_data)

conn.commit()

print("Database and students table ready!")
```

# Question 9 — Subquery + COUNT()

 
Find the courses that have more students than the average number of students per course.

```python
cursor.execute("""
SELECT Course, COUNT(*) AS Student_Count
FROM students
GROUP BY Course
HAVING COUNT(*) > (
    SELECT AVG(Student_Count)
    FROM (
        SELECT COUNT(*) AS Student_Count
        FROM students
        GROUP BY Course
    )
)
""")

result = cursor.fetchall()
print(result)
```

# Question 10 — Final Challenge 🔥

 
Display Name, Marks, and a Performance column. Use a subquery to find the average marks, then classify:

Marks > average → "Above Average"
Marks = average → "Average"
Marks < average → "Below Average"

```python
cursor.execute("""
SELECT Name, Marks,
       CASE
           WHEN Marks > (SELECT AVG(Marks) FROM students)
               THEN 'Above Average'
           WHEN Marks = (SELECT AVG(Marks) FROM students)
               THEN 'Average'
           ELSE 'Below Average'
       END AS Performance
FROM students
""")

result = cursor.fetchall()
print(result)
```
