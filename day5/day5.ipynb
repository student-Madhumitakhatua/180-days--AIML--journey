Problem 1 — Create a NumPy Array     

Create a NumPy array containing the numbers 5, 10, 15, 20, 25 and print the array.


```python
import numpy as np
arr=np.array([5,10,15,20,25])
print(arr)
```

Practice 2 — Check Array Properties          

For the array [10, 20, 30, 40, 50], print its number of dimensions, shape, and size.

```python
import numpy as np 
arr=np.array([10,20,30,40,50])
print(arr.size)
print(arr) #NORMAL PRINT 
print(arr.ndim) # DIMENSION
print(arr.shape) #SHAPE
```

Practice 3 — 2D Array   

Create a 2D NumPy array with 2 rows and 3 columns:


```python
import numpy as np
arr=np.array([
    [10,20,30],
    [40,50,60]
])
print(arr)
print(arr.ndim)
print(arr.shape)
print(arr.size)
```

Practice 4 — Array Indexing

```python
import numpy as np
arr=np.array([10,20,30,40,50])
print(arr[0])
print(arr[2])
print(arr[-1])
```

Practice 5 — 2D Array Indexing

```python
import numpy as np
arr=np.array([
    [10,20,30],
    [40,50,60]
])
print(arr[0] [0])
print(arr[1] [1])
print(arr[1][2])
```

Practice 6 — Slicing

```python
import numpy as np
arr=np.array([10,20,30,40,50])
print(arr[0:3])
print(arr[3:5])
print(arr[1:-1])
```

Practice 7 — Array Operations


Every element + 5

Every element × 2

Every element ÷ 10

Every element − 5

```python
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
print(arr+5)
print(arr*2)
print(arr/10)
print(arr-5)
```

Day 5 — NumPy Mini Assignment

```python
import numpy as np
#creating numpy array
arr = np.array([
    ["Madhu", 80],
    ["Punam", 70],
    ["Rahul", 90],
    ["Amit", 60],
    ["Priya", 85]
])
#basic operations
print(arr)
print(arr.ndim)
print(arr.shape)
print(arr.size)

#indexing
print(arr[0])
print(arr[2])
print(arr[4])

#slicing
print(arr[0:3])
print(arr[3:5])

#bonus mark
marks = np.array([80, 70, 90, 60, 85])
print(marks+5)
#average
print(np.mean(marks))



```
