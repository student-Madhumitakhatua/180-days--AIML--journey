🟢 Question 1

Create the following DataFrame and check which values are missing.



```python
import pandas as pd
data = {
    "Name": ["Madhu", "Punam", "Rahul", "Amit"],
    "Marks": [90, None, 85, None],
    "Age": [25, 20, None, 21]
}
df = pd.DataFrame(data)
df.isnull()
```

🟢 Question 2
 
Check how many missing values are present in each column.

```python
df.isnull().sum()
```

🟢 Question 3

Remove all rows that contain missing values.

```python
df.dropna()
```

🟢 Question 4

Fill all missing values in the Marks column with 0.

```python
df["Marks"].fillna(0)
```

🟢 Question 5

Fill the missing values in the Age column with the average age.

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

🟢 Question 6

Check again whether the DataFrame still contains any missing values.

```python
df.isnull().sum()
```

🟢 Question 7
 
Now permanently replace all missing values in the Marks column with 0.

```python
df["Marks"] = df["Marks"].fillna(0)
```

🟢 Question 8
 
Create a duplicate row in the DataFrame and then remove the duplicate rows.

```python
df = pd.concat([df, df.iloc[[3]]])
# df=df.drop_duplicates()
df
```

Check whether any duplicate rows are still present in the DataFrame.
