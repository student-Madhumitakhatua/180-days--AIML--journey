```python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6]
marks = [40, 50, 55, 65, 75, 85]

plt.scatter(hours, marks)
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.show()
```

🟢 Question 6 — Y-axis Label

 
Add a y-axis label: Marks

```python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6]
marks = [40, 50, 55, 65, 75, 85]

plt.scatter(hours, marks)
plt.title("Study Hours vs Marks")
plt.ylabel("marks")
plt.show()
```
