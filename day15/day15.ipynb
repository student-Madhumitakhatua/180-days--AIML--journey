```python
import sqlite3

conn = sqlite3.connect("day15.db")
cursor = conn.cursor()

print("Connected!")
```

```python
cursor.execute("""
CREATE TABLE students (
    Name TEXT,
    Age INTEGER,
    Marks INTEGER,
    Course TEXT
)
""")

conn.commit()

print("Table created!")
```

```python
students_data = [
    ("Madhu", 25, 90, "AI"),
    ("Punam", 20, 80, "Web"),
    ("Rahul", 22, 85, "ML"),
    ("Amit", 21, 70, "AI"),
    ("Priya", 23, 95, "ML")
]

cursor.executemany(
    "INSERT INTO students (Name, Age, Marks, Course) VALUES (?, ?, ?, ?)",
    students_data
)

conn.commit()

print("Data inserted!")
```

Display all columns and all students from the students table.

```python
cursor.execute("""
SELECT *
FROM students
""")

result = cursor.fetchall()
print(result)
```

Display only the Name and Marks columns.

```python
cursor.execute("""
SELECT Name, Marks
FROM students
""")
result = cursor.fetchall()
print(result)
```

##🟢 Q3 — WHERE

  Display students whose Marks are greater than 80.

```python
cursor.execute("""
SELECT *
FROM students
WHERE Marks > 80
""")

result = cursor.fetchall()
print(result)
```

##🟢 Q4 — ORDER BY
 

Display all students sorted by Marks from highest to lowest.

```python
cursor.execute("""
SELECT *
FROM students
ORDER BY  Marks DESC
""")

result = cursor.fetchall()
print(result)
```

##🟢 Q5 — LIMIT
 

Display only the top 3 students by Marks.

```python
cursor.execute("""
SELECT *
FROM students
ORDER BY Marks DESC
LIMIT 3
""")

result = cursor.fetchall()
print(result)
```

##🟢 Q6 — WHERE + ORDER BY
 

Display students whose Marks are 80 or more, sorted from highest to lowest Marks.

```python
cursor.execute("""
SELECT *
FROM students
WHERE Marks >= 80
ORDER BY Marks DESC
""")


result = cursor.fetchall()
print(result)
```

##🟢 Q7 — WHERE with Age
  Display students whose Age is greater than 21.

```python
cursor.execute("""
SELECT *
FROM students
WHERE Age > 21
""")

result = cursor.fetchall()
print(result)
```

##🟢 Q8 — WHERE + Specific Columns

  Display only Name and Course of students whose Marks are 85 or more.

```python
cursor.execute("""
SELECT Name, Course
FROM students
WHERE Marks >=85
""")

result = cursor.fetchall()
print(result)
```

##🟢 Q9 — LIMIT + ORDER BY

  Find the student with the lowest Marks.

```python
cursor.execute("""
SELECT *
FROM students
ORDER BY Marks ASC
LIMIT 1
""")

result = cursor.fetchall()
print(result)
```

##
🟢 Q10 — Final Challenge

 Find the average Marks of all students.

```python
cursor.execute("""
SELECT AVG(Marks)
FROM students
""")

result = cursor.fetchall()
print(result)
```
