# Question 1 — COUNT()

 
Find the total number of students in the students table.

```python
import sqlite3

conn = sqlite3.connect("day17.db")
cursor = conn.cursor()
```

```python
cursor.execute("""
CREATE TABLE students (
    Name TEXT,
    Age INTEGER,
    Marks INTEGER,
    Course TEXT
)
""")

conn.commit()
```

```python
students_data = [
    ("Madhu", 25, 90, "AI"),
    ("Punam", 20, 80, "Web"),
    ("Rahul", 22, 85, "ML"),
    ("Amit", 21, 70, "AI"),
    ("Priya", 23, 95, "ML")
]

cursor.executemany(
    "INSERT INTO students (Name, Age, Marks, Course) VALUES (?, ?, ?, ?)",
    students_data
)

conn.commit()
```

```python
cursor.execute("""
SELECT COUNT(*)
FROM students
""")

result = cursor.fetchone()
print(result)
```

# Question 2 — SUM()

 
Find the total sum of all students' Marks.

```python
cursor.execute("""
SELECT SUM(Marks)
FROM students
""")

result = cursor.fetchone()
print(result)
```

# Question 3 — AVG()

 
Find the average Marks of all students.

```python
cursor.execute("""
SELECT AVG(Marks)
FROM students
""")

result = cursor.fetchone()
print(result)
```



# Question 6 — GROUP BY

 
Find the average Marks for each Course.

```python
cursor.execute("""
SELECT Course, AVG(Marks)
FROM students
GROUP BY Course
""")

result = cursor.fetchall()
print(result)
```

# Question 7 — GROUP BY + COUNT()

 
Find the number of students in each Course.

```python
cursor.execute("""
SELECT Course, COUNT(*)
FROM students
GROUP BY Course
""")

result = cursor.fetchall()
print(result)
```

# Question 8 — GROUP BY + SUM()

 
Find the total Marks for each Course.

```python
cursor.execute("""
SELECT Course, SUM(Marks)
FROM students
GROUP BY Course
""")

result = cursor.fetchall()
print(result)
```

# Question 9 — HAVING

 
Find the courses whose total Marks are greater than 100.

```python
cursor.execute("""
SELECT Course, SUM(Marks)
FROM students
GROUP BY Course
HAVING SUM(Marks) > 100
""")

result = cursor.fetchall()
print(result)
```

 # Question 10 — Final Challenge 🔥

 
Find the average Marks for each Course, but show only courses whose average Marks is greater than 80.

```python
cursor.execute("""
SELECT Course, AVG(Marks)
FROM students
GROUP BY Course
HAVING AVG(Marks) > 80
""")

result = cursor.fetchall()
print(result)
```
