🚀 Day 14 — Mini Project: E-commerce Data Analysis

Step 1 — Dataset

```python
import pandas as pd

data = {
    "Product": ["Laptop", "Phone", "Headphones", "Keyboard", "Mouse",
                "Monitor", "Tablet", "Smartwatch", "Speaker", "Camera"],
    "Category": ["Electronics", "Electronics", "Audio", "Accessories", "Accessories",
                 "Electronics", "Electronics", "Wearable", "Audio", "Electronics"],
    "Price": [60000, 30000, 5000, 2000, 1000,
              15000, 25000, 8000, 4000, 45000],
    "Sales": [25, 40, 60, 80, 100,
              35, 30, 50, 70, 20]
}

df = pd.DataFrame(data)

df
```

Find the average Price of all products.

```python
df["Price"].mean()
```

🟢 Question 7

 
Find the average Sales of all products.

```python
df["Sales"].mean()
```

🟢 Question 8

 
Find all products where Sales > 50.

```python
df[df["Sales"] > 50]
```

🟢 Question 9
 
Create a bar chart showing Product vs Sales.

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(12, 6))

plt.bar(df["Product"], df["Sales"])

plt.xlabel("Product")
plt.ylabel("Sales")
plt.title("Product vs Sales")

plt.xticks(rotation=45)

plt.show()
```
