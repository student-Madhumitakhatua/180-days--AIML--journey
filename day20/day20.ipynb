```python
import sqlite3

conn = sqlite3.connect("day20.db")
cursor = conn.cursor()

print("Database connected!")
```

```python
cursor.execute("""
CREATE TABLE students (
    Name TEXT,
    Age INTEGER,
    Marks INTEGER,
    Course TEXT
)
""")

conn.commit()

print("Students table created!")
```

```python
students_data = [
    ("Madhu", 25, 90, "AI"),
    ("Punam", 20, 80, "Web"),
    ("Rahul", 22, 85, "ML"),
    ("Amit", 21, 70, "AI"),
    ("Priya", 23, 95, "ML")
]

cursor.executemany("""
INSERT INTO students (Name, Age, Marks, Course)
VALUES (?, ?, ?, ?)
""", students_data)

conn.commit()

print("Student data inserted!")
```

```python
cursor.execute("""
SELECT *
FROM students
""")

result = cursor.fetchall()
print(result)
```

# Question 1 — AVG() OVER()
 
Display Name, Marks, and the overall average marks beside every student.

```python
cursor.execute("""
SELECT
    Name,
    Marks,
    AVG(Marks) OVER() AS Overall_Average
FROM students
""")

result = cursor.fetchall()
print(result)
```

 # Question 2 — PARTITION BY

```python
cursor.execute("""
SELECT
    Name,
    Marks,
    AVG(Marks) OVER(PARTITION BY Course)
FROM students
""")

result = cursor.fetchall()
print(result)
```

# Question 3 — ROW_NUMBER()
 
Assign a row number to each student within their course, based on marks from highest to lowest.

```python
cursor.execute("""
SELECT
    Name,
    Marks,
    ROW_NUMBER() OVER(
    PARTITION BY Course
    ORDER BY Marks DESC
)
FROM students
""")

result = cursor.fetchall()
print(result)
```

# Question 4 — RANK()
 
Rank students within each course based on marks from highest to lowest.

```python
cursor.execute("""
SELECT
    Name,
    Marks,
    RANK() OVER(
    PARTITION BY Course
    ORDER BY Marks DESC
)
FROM students
""")

result = cursor.fetchall()
print(result)
```

# Question 5 — DENSE_RANK()
 
Rank students within each course based on marks from highest to lowest using DENSE_RANK().

```python
cursor.execute("""
SELECT
    Name,
    Marks,
    DENSE_RANK() OVER(
    PARTITION BY Course
    ORDER BY Marks DESC
)
FROM students
""")

result = cursor.fetchall()
print(result)
```

# Question 6 — LAG()
 
Show each student's marks and the previous student's marks when ordered by marks.

```python
cursor.execute("""
SELECT
    Name,
    Marks,
    LAG(Marks) OVER(
    ORDER BY Marks DESC
)
FROM students
""")

result = cursor.fetchall()
print(result)
```

# Question 7 — LEAD()
 
Show each student's marks and the next student's marks when ordered by marks from highest to lowest.

```python
cursor.execute("""
SELECT
    Name,
    Marks,
    LEAD(Marks) OVER(
    ORDER BY Marks DESC
)
FROM students
""")

result = cursor.fetchall()
print(result)
```

# Question 8 — SUM() OVER()
