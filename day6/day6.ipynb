Mini Assignment: Student Marks Analysis

```python
import numpy as np

marks = np.array([45, 67, 89, 32, 76, 90, 55, 82, 38, 95])

print(np.sum(marks))
print(np.mean(marks))
print(np.max(marks))
print(np.min(marks))
print(marks[marks>=70])
print(marks[marks<50])
print(marks[marks<50]+5)
marks[marks < 50] = marks[marks < 50] + 5
print(marks)
print(np.max(marks))
print(np.mean(marks[marks>80]))
 
```
