#count frequency(using dict,key value pairs)

```python
numbers = [1, 2, 2, 3, 3, 3, 4, 4]
d=dict()
for num in numbers:
    if num not in d:
        d[num]=1
    else:
        d[num]=d[num]+1  
        
    
        print(d)

```

```python
numbers = [1, 2, 2, 3, 3, 3, 4, 4]

d = {}

for num in numbers:
    d[num] = d.get(num, 0) + 1

print(d)
```

#Find Common Elements

```python
a = [1, 2, 3, 4, 5]
b = [4, 5, 6, 7, 8]
def find_common(a,b):
    result = []
    for num in a:
        if num in b and num not in result:
            result.append(num)
    return result
print(find_common(a,b))

    
```

```python
marks = [85, 72, 91, 68, 95]
total=sum(marks)
average=total/len(marks)
highest=max(marks)
lowest=min(marks)
print("Total marks:", total)
print("Average marks:", average)
print("Highest marks:", highest)
print("Lowest marks:", lowest)
```

#Problem 10 — Mini Challenge

```python
students = [
    {"name": "A", "score": 85},
    {"name": "B", "score": 92},
    {"name": "C", "score": 78},
    {"name": "D", "score": 95},
    {"name": "E", "score": 88}
]

average_score = sum(student["score"] for student in students) / len(students)

print(f"Average score: {average_score}")

highest_score_student=max(students, key=lambda x: x["score"])
print(f"highest score student name:{highest_score_student['name']}")

lowest_score_student=min(students, key=lambda x: x["score"])
print(f"lowest score student name:{lowest_score_student['name']}")

high_performers = [student for student in students if student["score"] >= 85]
print(f"High performers: {[student['name'] for student in high_performers]}")
```
