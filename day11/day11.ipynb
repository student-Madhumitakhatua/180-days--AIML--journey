🟢 Question 1

 
Create the following DataFrame and find the average marks for each department.

```python
import pandas as pd
data = {
    "Name": ["Madhu", "Punam", "Rahul", "Amit", "Priya"],
    "Department": ["AI", "Web", "AI", "Web", "AI"],
    "Marks": [90, 80, 85, 70, 95]
}
df=pd.DataFrame(data)
df.groupby("Department")["Marks"].mean()
```

🟢 Question 2

 
Find the total marks for each department.

```python
df.groupby("Department")["Marks"].sum()
```

🟢 Question 3

 
Find the highest marks in each department.

```python
df.groupby("Department")["Marks"].max()
```

🟢 Question 4

 
Find the lowest marks in each department.

```python
df.groupby("Department")["Marks"].min()
```

🟢 Question 5

 
Count how many students are present in each department.

```python
df.groupby("Department")["Name"].count()
```

🟢 Question 6

 
Find the average marks and maximum marks for each department.

```python
df.groupby("Department")["Marks"].agg(["mean", "max"])
```

🟢 Question 7

 
Find the total number of students in the DataFrame.

```python
df["Name"].count()
```

🟢 Question 8

 
Find the average marks of all students in the DataFrame.

```python
df["Marks"].mean()
```

🟢 Question 9

 
Find the total marks of students in the AI department.

```python
df[df["Department"] == "AI"]["Marks"].sum()
```

🟢 Question 10 — Final Practice

 
Find the department that has the highest average marks.

```python
avg = df.groupby("Department")["Marks"].mean()
 
avg.idxmax()
 
```

```python
import pandas as pd

data = {
    "Name": ["Madhu", "Punam", "Rahul", "Amit", "Priya"],
    "Department": ["AI", "Web", "AI", "Web", "AI"],
    "Marks": [90, 80, 85, 70, 95]
}

df = pd.DataFrame(data)

# 1. Average marks for each department
df.groupby("Department")["Marks"].mean()

# 2. Total marks for each department
df.groupby("Department")["Marks"].sum()

# 3. Highest marks in each department
df.groupby("Department")["Marks"].max()

# 4. Lowest marks in each department
df.groupby("Department")["Marks"].min()

# 5. Number of students in each department
df.groupby("Department")["Name"].count()

# 6. Average and maximum marks
df.groupby("Department")["Marks"].agg(["mean", "max"])

# 7. Total number of students
df["Name"].count()

# 8. Overall average marks
df["Marks"].mean()

# 9. Total marks of AI department
df[df["Department"] == "AI"]["Marks"].sum()

# 10. Department with highest average marks
avg = df.groupby("Department")["Marks"].mean()
avg.idxmax()
```
