🟢 Question 1


Create these two DataFrames and merge them using the common Name column.

```python
import pandas as pd
students = pd.DataFrame({
    "Name": ["Madhu", "Punam", "Rahul"],
    "Age": [25, 20, 22]
})

marks = pd.DataFrame({
    "Name": ["Madhu", "Punam", "Rahul"],
    "Marks": [90, 80, 85]
})
pd.merge(students, marks, on="Name")
```

🟢 Question 2

 
Sort the merged DataFrame by Age from youngest to oldest.

```python

df = pd.merge(students, marks, on="Name")
df.sort_values("Age", ascending=True)
```

🟢 Question 3
 
Sort the DataFrame by Marks from highest to lowest.

```python
df.sort_values("Marks", ascending=False)
```
