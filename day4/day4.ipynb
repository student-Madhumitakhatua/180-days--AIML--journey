Day 04 — Mini Challenge    

Create a student JSON file and write a Python program that safely reads it, displays the student's information, and handles the case when the file does not exist.

```python
import json as js
data={
    "name": "Madhu",
    "age": 25,
    "course": "BCA",
    "marks": 90
}
with open("student.json", "w") as file:
    js.dump(data, file)
    
try:
    with open ("student.json","r") as file:
        data=js.load(file)
        print("Name:", data["name"])
        print("Age:", data["age"])
        print("course:", data["course"])
        print("marks:", data["marks"])

except FileNotFoundError:
    print(" file not found")
finally:
    print("program finish")
```
