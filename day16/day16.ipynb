3📝 Question 1 — WHERE

 
Write an SQL query to display students whose marks are greater than 80.

```python
import sqlite3

conn = sqlite3.connect("day15.db")
cursor = conn.cursor()

print("Database connected!")
```

```python
cursor.execute("""
CREATE TABLE students (
    Name TEXT,
    Age INTEGER,
    Marks INTEGER,
    Course TEXT
)
""")

conn.commit()
```

```python
students_data = [
    ("Madhu", 25, 90, "AI"),
    ("Punam", 20, 80, "Web"),
    ("Rahul", 22, 85, "ML"),
    ("Amit", 21, 70, "AI"),
    ("Priya", 23, 95, "ML")
]

cursor.executemany(
    "INSERT INTO students (Name, Age, Marks, Course) VALUES (?, ?, ?, ?)",
    students_data
)

conn.commit()
```

```python
cursor.execute("SELECT * FROM students")
print(cursor.fetchall())
```

```python
cursor.execute("""
SELECT *
FROM students
WHERE Marks > 80
""")

result = cursor.fetchall()
print(result)
```

#📝 Question 2 — AND

 
Display students whose Marks are greater than 80 AND Age is greater than 21.

```python
cursor.execute("""
SELECT *
FROM students
WHERE Marks > 80 AND    Age >21
""")

result = cursor.fetchall()
print(result)
```

#📝 Question 3 — OR

 
Display students whose Marks are greater than 90 OR Age is less than 21.

```python
cursor.execute("""
SELECT *
FROM students
WHERE Marks > 90 OR    Age <21
""")

result = cursor.fetchall()
print(result)
```

#📝 Question 4 — NOT

 
Display students whose Course is NOT "AI".

```python
cursor.execute("""
SELECT *
FROM students
WHERE NOT Course = "AI"
""")

result = cursor.fetchall()
print(result)
```

#📝 Question 5 — BETWEEN

 
Display students whose Marks are between 80 and 90.

```python
cursor.execute("""
SELECT *
FROM students
WHERE Marks BETWEEN 80 AND 90
""")

result = cursor.fetchall()
print(result)
```

#📝 Question 6 — IN

 
Display students whose Course is either "AI" or "ML".

```python
cursor.execute("""
SELECT *
FROM students
WHERE Course IN ("AI", "ML")
""")

result = cursor.fetchall()
print(result)
```

#📝 Question 7 — LIKE

 
Display students whose Name starts with "P".

 

```python
cursor.execute("""
SELECT *
FROM students
WHERE Name LIKE "P%"
""")

result = cursor.fetchall()
print(result)
```

#📝 Question 8 — LIKE + %

 
Display students whose Name ends with "a".

```python
cursor.execute("""
SELECT *
FROM students
WHERE Name LIKE "%a"
""")

result = cursor.fetchall()
print(result)
```

#📝 Question 9 — Combined Conditions

 
Display students whose Marks are greater than 80 AND Course is either AI or ML.

```python
cursor.execute("""
SELECT *
FROM students
WHERE Marks > 80 AND Course IN ("AI", "ML")
""")

result = cursor.fetchall()
print(result)
```

#📝 Question 10 — Final Challenge 🔥

 
Find students whose Marks are between 80 and 95 AND Course is not "Web".

```python
cursor.execute("""
SELECT *
FROM students
WHERE Marks BETWEEN 80 AND 95
AND NOT Course = "Web"
""")

result = cursor.fetchall()
print(result)
```
