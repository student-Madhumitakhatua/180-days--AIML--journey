```python
import sqlite3

conn = sqlite3.connect("day21.db")
cursor = conn.cursor()

print("Database connected!")
```

```python
cursor.execute("""
CREATE TABLE students (
    Name TEXT,
    Age INTEGER,
    Marks INTEGER,
    Course TEXT
)
""")

conn.commit()

print("Students table created!")
```

```python
students_data = [
    ("Madhu", 25, 90, "AI"),
    ("Punam", 20, 80, "Web"),
    ("Rahul", 22, 85, "ML"),
    ("Amit", 21, 70, "AI"),
    ("Priya", 23, 95, "ML")
]

cursor.executemany("""
INSERT INTO students (Name, Age, Marks, Course)
VALUES (?, ?, ?, ?)
""", students_data)

conn.commit()

print("Student data inserted!")
```

```python
cursor.execute("""
SELECT *
FROM students
""")

result = cursor.fetchall()
print(result)
```

```python
cursor.execute("""
CREATE TABLE courses (
    Course TEXT,
    Duration INTEGER
)
""")

conn.commit()

print("Courses table created!")
```

```python
courses_data = [
    ("AI", 6),
    ("ML", 8),
    ("Web", 5)
]

cursor.executemany("""
INSERT INTO courses (Course, Duration)
VALUES (?, ?)
""", courses_data)

conn.commit()

print("Course data inserted!")
```

```python
cursor.execute("""
SELECT *
FROM courses
""")

result = cursor.fetchall()
print(result)
```

# Question 1 — Filtering Revision
 
Display the Name, Marks, and Course of students whose Marks are greater than 80 and whose Course is either AI or ML.

```python
cursor.execute("""
SELECT Name, Marks, Course
FROM students
WHERE Marks > 80
AND Course IN ('AI', 'ML')
""")

result = cursor.fetchall()
print(result)
```

 # Question 2 — GROUP BY
 
Find the average marks for each Course.

```python
cursor.execute("""
SELECT Course, AVG(Marks)
FROM students
GROUP BY Course
""")

result = cursor.fetchall()
print(result)
```

# Question 3 — HAVING
 
Display only the courses whose average marks are greater than 80.

```python
cursor.execute("""
SELECT Course, AVG(Marks)
FROM students
GROUP BY Course
HAVING AVG(Marks) > 80
""")

result = cursor.fetchall()
print(result)
```

# Question 4 — COUNT() + GROUP BY
 
Find the number of students in each Course.

```python
cursor.execute("""
SELECT Course, COUNT(*)
FROM students
GROUP BY Course
""")

result = cursor.fetchall()
print(result)
```

# Question 5 — MAX() + GROUP BY
 
Find the highest marks in each Course.

```python
cursor.execute("""
SELECT Course, MAX(Marks)
FROM students
GROUP BY Course
 
""")

result = cursor.fetchall()
print(result)
```

# Question 6 — JOIN Revision
 
Display each student's Name, Course, and Duration by joining the students table with the courses table.

```python
cursor.execute("""
SELECT students.Name, students.Course, courses.Duration
FROM students
INNER JOIN courses
ON students.Course = courses.Course
""")

result = cursor.fetchall()
print(result)
```

# Question 7 — JOIN + WHERE
 
Display the Name, Course, and Duration of students whose course duration is greater than 5 months.

```python
cursor.execute("""
SELECT students.Name, students.Course, courses.Duration
FROM students
INNER JOIN courses
ON students.Course = courses.Course
WHERE courses.Duration > 5
""")

result = cursor.fetchall()
print(result)
```

# Question 8 — Subquery Revision 🔥
 
Find students whose marks are greater than the overall average marks.

```python
cursor.execute("""
SELECT Name, Marks
FROM students
WHERE Marks > (
    SELECT AVG(Marks)
    FROM students
)
""")

result = cursor.fetchall()
print(result)
```

# Question 9 — CASE WHEN Revision
 
Display Name, Marks, and a new column Grade:

- Marks ≥ 90 → A

- Marks ≥ 80 → B

- Marks ≥ 70 → C

- Otherwise → D

```python
cursor.execute("""
SELECT Name, Marks,
       CASE
           WHEN Marks >= 90 THEN 'A'
           WHEN Marks >= 80 THEN 'B'
           WHEN Marks >= 70 THEN 'C'
           ELSE 'D'
       END AS Grade
FROM students
""")

result = cursor.fetchall()
print(result)
```

# Question 10 — Final Revision Challenge 🔥
 
Display each student's Name, Marks, and their rank within their Course, with the highest marks getting rank 1.

```python
cursor.execute("""
SELECT
    Name,
    Marks,
    RANK() OVER(
        PARTITION BY Course
        ORDER BY Marks DESC
    ) AS Rank
FROM students
""")

result = cursor.fetchall()
print(result)
```
