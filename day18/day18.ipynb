```python
import sqlite3

conn = sqlite3.connect("day18.db")
cursor = conn.cursor()

print("Database connected!")
```

```python
cursor.execute("""
CREATE TABLE students (
    Name TEXT,
    Age INTEGER,
    Marks INTEGER,
    Course TEXT
)
""")

conn.commit()

print("Students table created!")
```

```python
students_data = [
    ("Madhu", 25, 90, "AI"),
    ("Punam", 20, 80, "Web"),
    ("Rahul", 22, 85, "ML"),
    ("Amit", 21, 70, "AI"),
    ("Priya", 23, 95, "ML")
]

cursor.executemany(
    "INSERT INTO students (Name, Age, Marks, Course) VALUES (?, ?, ?, ?)",
    students_data
)

conn.commit()

print("Students data inserted!")
```

```python
cursor.execute("""
CREATE TABLE courses (
    Course TEXT,
    Duration INTEGER
)
""")

conn.commit()

print("Courses table created!")
```

```python
cursor.execute("SELECT * FROM courses")
result = cursor.fetchall()
print(result)
```

# Question 1 — INNER JOIN

 
Join the students and courses tables using the Course column and display:

```python
cursor.execute("""
SELECT students.Name, students.Course, courses.Duration
FROM students
INNER JOIN courses
ON students.Course = courses.Course
""")

result = cursor.fetchall()
print(result)
```

 # Question 2 — INNER JOIN

 
Using INNER JOIN, display student Name and Course Duration only.

```python
cursor.execute("""
SELECT students.Name, courses.Duration
FROM students
INNER JOIN courses
ON students.Course = courses.Course
""")

result = cursor.fetchall()
print(result)
```

 # Question 3 — INNER JOIN + WHERE

 
Display students whose Course Duration is greater than 5 months.

```python
cursor.execute("""
SELECT students.Name, students.Course
FROM students
INNER JOIN courses
ON students.Course = courses.Course
WHERE courses.Duration > 5
""")

result = cursor.fetchall()
print(result)
```

 # Question 4 — LEFT JOIN
 
Use LEFT JOIN to display all students, along with their course duration.

```python
cursor.execute("""
SELECT students.Name, courses.Duration
FROM students
LEFT JOIN courses
ON students.Course = courses.Course
""")

result = cursor.fetchall()
print(result)
```

 # Question 5 — LEFT JOIN + WHERE

 
Using LEFT JOIN, display the Name and Duration of students whose course duration is 8

```python
cursor.execute("""
SELECT students.Name, courses.Duration
FROM students
LEFT JOIN courses
ON students.Course = courses.Course
WHERE courses.Duration = 8
""")

result = cursor.fetchall()
print(result)
```

 # Question 6 — INNER JOIN + Sorting
 
Display Name, Course, and Duration, and sort the result by Duration from highest to lowest.

```python
cursor.execute("""
SELECT students.Name, students.Course, courses.Duration
FROM students
INNER JOIN courses
ON students.Course = courses.Course
ORDER BY courses.Duration DESC
""")

result = cursor.fetchall()
print(result)
```

 # Question 7 — LEFT JOIN + GROUP BY

 
Find the number of students in each course, using LEFT JOIN.

```python
cursor.execute("""
SELECT courses.Course, COUNT(students.Name)
FROM courses
LEFT JOIN students
ON courses.Course = students.Course
GROUP BY courses.Course
""")

result = cursor.fetchall()
print(result)
```

 # Question 8 — INNER JOIN + WHERE + Sorting

 
Display Name, Course, and Duration for students whose course duration is greater than 5, sorted by Duration from highest to lowest.

```python
cursor.execute("""
SELECT students.Name, students.Course, courses.Duration
FROM students
INNER JOIN courses
ON students.Course = courses.Course
WHERE courses.Duration > 5
ORDER BY courses.Duration DESC
""")

result = cursor.fetchall()
print(result)
```

 # Question 9 — LEFT JOIN + ORDER BY

 
Display all students' Name and Course Duration, sorted by Duration from lowest to highest.

```python
cursor.execute("""
SELECT students.Name, students.Course, courses.Duration
FROM students
LEFT JOIN courses
ON students.Course = courses.Course
ORDER BY courses.Duration ASC
""")

result = cursor.fetchall()
print(result)
```

 # Question 10 — Final SQL JOIN Challenge 🔥

 
Display the Course name and the number of students enrolled in each course. Show only courses having more than 1 student.

```python
cursor.execute("""
SELECT courses.Course, COUNT(students.Name)
FROM courses
LEFT JOIN students
ON courses.Course = students.Course
GROUP BY courses.Course
HAVING COUNT(students.Name) > 1
""")

result = cursor.fetchall()
print(result)
```
